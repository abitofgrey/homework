```python
import pandas as pd
url = "https://raw.githubusercontent.com/nkmwicz/data-for-students/refs/heads/main/french-ambs.csv"
df = pd.read_csv(url)
```

```python
df.shape
```

```python
df[(df["year"] > 1588) & (df["year"] < 1592)].count()
```

```python
df["is_permenant"] = True
not_permenant = ["Geneva", "Saxony", "Tuscany", "Ferrara"]
for item in df.itertuples():
    if item.place in not_permenant:
        df.at[item.Index, "is_permenant"] = False
df[df["is_permenant"] == False].count()
```

```python
replace = ["Geneva", "Grisons"]
for item in df.itertuples():
    if item.place in replace:
        df.at[item.Index, "place"] = "Swiss Cantons"
df[df["place"] == "Swiss Cantons"].count()
```

```python
df_agg = df.groupby(["place"]).agg(
    years_present=("year", "count"),
    unique_ambs=("name", "nunique")
).sort_values(["unique_ambs"], ascending=True)
df_agg
```
